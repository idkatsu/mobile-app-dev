```polyglot-notebook
// Функция для рисования прямоугольника
public string[] GetRectangle(int width, int height, char symbol = '*', bool fill = true)
{
    if (width <= 0 || height <= 0) return new string[0];
    string[] result = new string[height];
    
    for (int i = 0; i < height; i++)
    {
        if (fill || i == 0 || i == height - 1)
        {
            result[i] = new string(symbol, width);
        }
        else
        {
            if (width == 1)
                result[i] = symbol.ToString();
            else
                result[i] = symbol + new string(' ', width - 2) + symbol;
        }
    }
    return result;
}

public void PrintRectangle(int width, int height, char symbol = '*', bool fill = true)
{
    foreach (var line in GetRectangle(width, height, symbol, fill))
        Console.WriteLine(line);
}

Console.WriteLine("=== ПРЯМОУГОЛЬНИК (заполненный, 10x4) ===");
PrintRectangle(10, 4, '*');
Console.WriteLine();

Console.WriteLine("=== ПРЯМОУГОЛЬНИК (полый, 10x5) ===");
PrintRectangle(10, 5, '#', fill: false);
Console.WriteLine();

Console.WriteLine("=== ПРЯМОУГОЛЬНИК (маленький, 1x1) ===");
PrintRectangle(1, 1, '+');
Console.WriteLine();
```

```polyglot-notebook
// Функция для рисования равнобедренного прямоугольного треугольника
public string[] GetIsoscelesRightTriangle(int size, char symbol = '*', bool pointDown = false)
{
    string[] result = new string[size];
    
    for (int i = 0; i < size; i++)
    {
        int row = pointDown ? size - 1 - i : i;
        int stars = row + 1;
        int spaces = size - stars;
        result[i] = new string(' ', spaces) + new string(symbol, stars);
    }
    return result;
}

public void PrintIsoscelesRightTriangle(int size, char symbol = '*', bool pointDown = false)
{
    foreach (var line in GetIsoscelesRightTriangle(size, symbol, pointDown))
        Console.WriteLine(line);
}

Console.WriteLine("=== ТРЕУГОЛЬНИК (указатель вверх, 6) ===");
PrintIsoscelesRightTriangle(6, '*', pointDown: false);
Console.WriteLine();

Console.WriteLine("=== ТРЕУГОЛЬНИК (указатель вниз, 5) ===");
PrintIsoscelesRightTriangle(5, '#', pointDown: true);
Console.WriteLine();
```

```polyglot-notebook
// Функция для рисования песочных часов
public string[] GetHourglass(int size, char symbol = '*')
{
    if (size % 2 == 0) size++; 
    
    string[] result = new string[size];
    int mid = size / 2;
    
    for (int i = 0; i < size; i++)
    {
        int stars;
        int spaces;
        
        if (i <= mid)
        {
            stars = size - 2 * i;
            spaces = i;
        }
        else
        {
            int dist = size - 1 - i;
            stars = size - 2 * dist;
            spaces = dist;
        }
        
        result[i] = new string(' ', spaces) + new string(symbol, stars);
    }
    return result;
}

public void PrintHourglass(int size, char symbol = '*')
{
    foreach (var line in GetHourglass(size, symbol))
        Console.WriteLine(line);
}

Console.WriteLine("=== ПЕСОЧНЫЕ ЧАСЫ (7x7) ===");
PrintHourglass(7, 'X');
Console.WriteLine();

Console.WriteLine("=== ПЕСОЧНЫЕ ЧАСЫ (9x9) ===");
PrintHourglass(9, '@');
Console.WriteLine();
```

```polyglot-notebook
public string[] GetZebra(int width, int height, int stripeThickness = 1, bool isHorizontal = true, char symbol = '*', char bgSymbol = ' ')
{
    string[] result = new string[height];
    for (int i = 0; i < height; i++)
    {
        if (isHorizontal)
        {
            bool isStripe = (i / stripeThickness) % 2 == 0;
            result[i] = new string(isStripe ? symbol : bgSymbol, width);
        }
        else
        {
            char[] row = new char[width];
            for (int j = 0; j < width; j++)
            {
                bool isStripe = (j / stripeThickness) % 2 == 0;
                row[j] = isStripe ? symbol : bgSymbol;
            }
            result[i] = new string(row);
        }
    }
    return result;
}

public void PrintZebra(int width, int height, int stripeThickness = 1, bool isHorizontal = true, char symbol = '*', char bgSymbol = ' ') =>
    Array.ForEach(GetZebra(width, height, stripeThickness, isHorizontal, symbol, bgSymbol), Console.WriteLine);

Console.WriteLine("\n=== ЗЕБРА (горизонтальная, толщина 2) ===");
PrintZebra(15, 8, stripeThickness: 2, isHorizontal: true, symbol: '@', bgSymbol: ' ');

Console.WriteLine("\n=== ЗЕБРА (вертикальная, толщина 3) ===");
PrintZebra(20, 6, stripeThickness: 3, isHorizontal: false, symbol: '#', bgSymbol: ' ');
```

```polyglot-notebook
public string[] CanvasToStrings(char[,] canvas)
{
    int height = canvas.GetLength(0);
    int width = canvas.GetLength(1);
    string[] result = new string[height];
    for (int i = 0; i < height; i++)
    {
        char[] row = new char[width];
        for (int j = 0; j < width; j++) row[j] = canvas[i, j];
        result[i] = new string(row);
    }
    return result;
}

public string[] GetNestedRectangles(int width, int height, char symbol = '*')
{
    char[,] canvas = new char[height, width];
    for (int r = 0; r < height; r++) 
        for (int c = 0; c < width; c++) 
            canvas[r, c] = ' ';
    
    for (int d = 0; d <= Math.Min(width, height) / 2; d += 2)
    {
        int top = d, bottom = height - 1 - d;
        int left = d, right = width - 1 - d;
        
        if (left > right || top > bottom) break;
        
        for (int x = left; x <= right; x++)
        {
            canvas[top, x] = symbol;
            canvas[bottom, x] = symbol;
        }
        for (int y = top; y <= bottom; y++)
        {
            canvas[y, left] = symbol;
            canvas[y, right] = symbol;
        }
    }
    return CanvasToStrings(canvas);
}

public void PrintNestedRectangles(int width, int height, char symbol = '*') =>
    Array.ForEach(GetNestedRectangles(width, height, symbol), Console.WriteLine);


Console.WriteLine("\n=== ВЛОЖЕННЫЕ ПРЯМОУГОЛЬНИКИ ===");
PrintNestedRectangles(25, 11, '@');
```

```polyglot-notebook
public string[] GetSpiral(int width, int height, bool clockwise = true, char symbol = '*')
{
    char[,] canvas = new char[height, width];
    for (int r = 0; r < height; r++) 
        for (int c = 0; c < width; c++) 
            canvas[r, c] = ' ';
    
    int[] dx = clockwise ? new[] { 1, 0, -1, 0 } : new[] { 0, 1, 0, -1 };
    int[] dy = clockwise ? new[] { 0, 1, 0, -1 } : new[] { 1, 0, -1, 0 };
    
    int x = 0, y = 0, dir = 0;
    canvas[y, x] = symbol;
    
    bool CanMove(int cx, int cy, int cdir)
    {
        int nx = cx + dx[cdir];
        int ny = cy + dy[cdir];
        
        if (nx < 0 || nx >= width || ny < 0 || ny >= height) return false;
        if (canvas[ny, nx] == symbol) return false;
        
        int neighbors = 0;
        if (ny > 0 && canvas[ny - 1, nx] == symbol) neighbors++;
        if (ny < height - 1 && canvas[ny + 1, nx] == symbol) neighbors++;
        if (nx > 0 && canvas[ny, nx - 1] == symbol) neighbors++;
        if (nx < width - 1 && canvas[ny, nx + 1] == symbol) neighbors++;
        
        return neighbors <= 1;
    }
    
    while (true)
    {
        if (CanMove(x, y, dir))
        {
            x += dx[dir];
            y += dy[dir];
            canvas[y, x] = symbol;
        }
        else
        {
            dir = (dir + 1) % 4;
            if (CanMove(x, y, dir))
            {
                x += dx[dir];
                y += dy[dir];
                canvas[y, x] = symbol;
            }
            else break;
        }
    }
    return CanvasToStrings(canvas);
}

public void PrintSpiral(int width, int height, bool clockwise = true, char symbol = '*') =>
    Array.ForEach(GetSpiral(width, height, clockwise, symbol), Console.WriteLine);


Console.WriteLine("\n=== УЛИТКА (по часовой) ===");
PrintSpiral(20, 9, clockwise: true, symbol: '#');

Console.WriteLine("\n=== УЛИТКА (против часовой) ===");
PrintSpiral(20, 9, clockwise: false, symbol: '#');  
```

```polyglot-notebook
public string[] GetSnake(int width, int height, bool isHorizontal = true, int thickness = 1, bool directionNormal = true, char symbol = '*')
{
    char[,] canvas = new char[height, width];
    for (int r = 0; r < height; r++) 
        for (int c = 0; c < width; c++) 
            canvas[r, c] = ' ';
    
    for (int y = 0; y < height; y++)
    {
        for (int x = 0; x < width; x++)
        {
            if (isHorizontal)
            {
                int phase = y / thickness;
                if (phase % 2 == 0) 
                    canvas[y, x] = symbol;
                else
                {
                    bool connectRight = directionNormal ? (phase % 4 == 1) : (phase % 4 == 3);
                    if (connectRight && x >= width - thickness) canvas[y, x] = symbol;
                    if (!connectRight && x < thickness) canvas[y, x] = symbol;
                }
            }
            else
            {
                int phase = x / thickness;
                if (phase % 2 == 0) 
                    canvas[y, x] = symbol;
                else
                {
                    bool connectBottom = directionNormal ? (phase % 4 == 1) : (phase % 4 == 3);
                    if (connectBottom && y >= height - thickness) canvas[y, x] = symbol;
                    if (!connectBottom && y < thickness) canvas[y, x] = symbol;
                }
            }
        }
    }
    return CanvasToStrings(canvas);
}

public void PrintSnake(int width, int height, bool isHorizontal = true, int thickness = 1, bool directionNormal = true, char symbol = '*') =>
    Array.ForEach(GetSnake(width, height, isHorizontal, thickness, directionNormal, symbol), Console.WriteLine);

Console.WriteLine("\n=== ЗМЕЙКА (горизонтальная, толщина 2) ===");
PrintSnake(15, 10, isHorizontal: true, thickness: 2, directionNormal: true, symbol: '*');

Console.WriteLine("\n=== ЗМЕЙКА (вертикальная, толщина 2) ===");
PrintSnake(20, 9, isHorizontal: false, thickness: 2, directionNormal: true, symbol: '*');
```
